```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
import lightgbm as lgb
```

```python
colunas = ['preg', 'plas', 'pres', 'skin', 'test', 'mass', 'pedi', 'age', 'class']
dados = pd.read_csv('pima-data.csv', names=colunas)
dados.head()
```

```python
array = dados.values

X = array[:, 0:8]
y = array[:, 8]
```

```python
test_size = 0.33
seed = 7

Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=test_size, random_state=seed)
```

```python
modelo = lgb.LGBMClassifier(learning_rate = 0.01, n_estimators=300, max_depth=4)

modelo.fit(Xtrain, ytrain)
ypred = modelo.predict(Xtest)

previsoes = [round(value) for value in ypred]

print('Acurácia: {:.2f}'.format(accuracy_score(ytest, previsoes) * 100))
```

```python
from xgboost import plot_importance
import matplotlib.pyplot as plt

plot_importance(modelo)
plt.show()
```
