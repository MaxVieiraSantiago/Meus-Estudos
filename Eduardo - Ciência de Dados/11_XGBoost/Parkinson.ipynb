```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
from sklearn.preprocessing import MinMaxScaler
```

```python
dados = pd.read_csv('parkinsons.data')
dados.head()
```

```python
dados.isnull().sum()
```

```python
dados.info()
```

```python
X = dados.drop(columns=['name', 'status'], axis=1)
y = dados['status']

scaler = MinMaxScaler((-1,1))
X_scaler = scaler.fit_transform(X)

Xtrain, Xtest, ytrain, ytest = train_test_split(X_scaler, y, test_size=0.2, random_state=7)
```

```python
labels = y
print(labels[labels == 1].shape[0], labels[labels == 0].shape[0])
```

```python
pd.DataFrame(X_scaler, columns=X.columns).head()
```

```python
modelo = XGBClassifier(learning_rate=0.2, n_estimators=500, max_depth=15)

modelo.fit(Xtrain, ytrain)
ypred = modelo.predict(Xtest)

print('Acurácia: {:.2f}%'.format(accuracy_score(ytest, ypred) * 100))
```
