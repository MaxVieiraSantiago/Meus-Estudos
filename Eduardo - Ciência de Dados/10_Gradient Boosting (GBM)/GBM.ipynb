## **Gradient Boosting**

https://www.youtube.com/watch?v=jPuT_bYClQw

Também chamado de **Stochastic Gradient Boosting**, é um dos métodos Ensemble mas sofisticados.

Um algoritmo ensemble combina os resultados de múltiplos modelos de machine learning para melhorar o desempenho geral. Em vez de depender de um único modelo, a técnica utiliza vários modelos (chamados de base learners) para reduzir erros como viés ou variância. Os dois principais tipos de ensembles são:

* **Bagging (Bootstrap Aggregating)**: Treina vários modelos em subconjuntos aleatórios dos dados (ex.: **Random Forest**).
* **Boosting**: Treina modelos sequencialmente, ajustando erros do modelo anterior (ex.: **Gradient Boosting**).


A ideia central é que a combinação de modelos melhora a precisão e a robustez em relação a modelos individuais.

```python
# Importando os módulos
import pandas as pd

from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import GradientBoostingClassifier

# Importando os dados
arquivo = 'pima-data.csv'

dados = pd.read_csv( arquivo)
dados.head()

# Separando o array em componentes de entrada e saída
X = dados.drop(columns=['diabetes'], axis=1)
y = dados['diabetes']

# Definindo os valores para os números de fold
num_folds = 10
seed = 7

# Definindo  número de trees
num_trees = 100

# Separando os dados em folds
kfold = KFold(n_splits=num_folds, shuffle=True, random_state=seed)

# Criando o modelo
modelo = GradientBoostingClassifier(n_estimators=num_trees, random_state=seed)

# Cross Validation
resultado = cross_val_score(modelo, X, y, cv=kfold)

# Print do resultado
print("Acurácia: {:.3f}".format(resultado.mean() * 100))
```
