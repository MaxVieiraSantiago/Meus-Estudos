## **Redes Neurais**

Dataset:
https://archive.ics.uci.edu/dataset/165/concrete+compressive+strength

Modelagem da resistência do concreto.

```python
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import Normalizer
from sklearn.preprocessing import MinMaxScaler

from sklearn.neural_network import MLPRegressor

from sklearn.model_selection import train_test_split

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
dados = pd.read_csv('Concrete_Data.csv', sep=';')
dados.head()
```

```python
dados.info()
```

```python
for col in dados.columns:
    if dados[col].dtypes == 'object':
        dados[col] = dados[col].str.replace(',', '.').astype(float)
    
dados.head()
```

```python
dados.info()
```

```python
normalizer = MinMaxScaler(feature_range=(0, 1))

dados = pd.DataFrame({col: normalizer.fit_transform(dados[[col]])[:, 0] for col in dados.columns})
dados.head()
```

```python
X = dados.drop(columns=['Concrete Strength'])
y = dados['Concrete Strength']

Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.30)
```

```python
neural = MLPRegressor(hidden_layer_sizes=(5))

neural.fit(Xtrain, ytrain)

ypred = neural.predict(Xtest)
```

```python
print(mean_squared_error(ytest, ypred))
print(mean_absolute_error(ytest, ypred))
```
