## **Normalização ou Padronização**

**Normalizar**

X' = ( X - Xmin ) / ( Xmax - Xmin )

Valores São Normalizados entre ( 0, 1 )

**Padronizar ou Z-Score Normalizaton** - https://www.youtube.com/watch?v=-LC_PKBoZfk

**Mean / Standard Deviation**\
X' = ( X - Xmédia ) / DesvPad  =>  Média fica igual a 0 e o Desvio Padrão igual a 1

**Mean / Mean Absolute Deviation**\
Mean Absolute Deviation = **A**\
**A** = ( ( X1 - Xmédia ) + ( X2 - Xmédia ) + ... ( Xn - Xmédia ) )\
X1 = ( X - Xmédia ) / A

## **Resultado da Máquina Preditiva Sem Redimensionamento**

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Carga dos Dados
dados = pd.read_csv( 'pima-data.csv')
```

```python
# Separando os dados
X = dados.drop(columns=['diabetes'], axis=1)
y = dados['diabetes']
```

```python
# Definindo os Valores para os Números de Folds
num_folds = 10
random_state = 7
```

```python
# Separando os dados em Folds
kfold = KFold(n_splits=num_folds, shuffle=True, random_state=random_state)
```

```python
# Criando o modelo
modelo = KNeighborsClassifier()

# Cross Validation
resultados = cross_val_score(modelo, X, y, cv=kfold)

# Resultado
print('Acurácia: %.3f' % (resultados.mean() * 100))
```

## **Resultado com Padronização**

```python
scaler = StandardScaler()
X_scaler = scaler.fit_transform(X)

# Criando o modelo
modelo = KNeighborsClassifier()

# Cross Validation
resultados = cross_val_score(modelo, X_scaler, y, cv=kfold)

# Resultado
print('Acurácia: %.3f' % (resultados.mean() * 100))
```

## **Resultado com Normalização**

```python
scaler = MinMaxScaler(feature_range=(0,1))
X_scaler = scaler.fit_transform(X)

# Criando o modelo
modelo = KNeighborsClassifier()

# Cross Validation
resultados = cross_val_score(modelo, X_scaler, y, cv=kfold)

# Resultado
print('Acurácia: %.3f' % (resultados.mean() * 100))
```

## **Z-Score - Normalization ( Mean / Mean Absolut Deviation )**

```python
# Carga dos Dados
dados = pd.read_csv( 'pima-data.csv')
dados.head()
```

```python
import numpy as np

def z_score_mad(data):
    mean = np.mean(data)
    mad = np.mean(np.abs(data - mean))  # MAD
    z_scores = (data - mean) / mad
    return z_scores
```

```python
colunas = dados.columns.drop('diabetes')
z_score = z_score_mad(dados[colunas])
z_score['diabetes'] = dados['diabetes']

z_score.head()
```

```python
X = z_score.drop(columns=['diabetes'], axis=1)
y = z_score['diabetes']

modelo = KNeighborsClassifier()

# Cross Validation
resultados = cross_val_score(modelo, X, y, cv=kfold)

# Resultado
print('Acurácia: %.3f' % (resultados.mean() * 100))
```
