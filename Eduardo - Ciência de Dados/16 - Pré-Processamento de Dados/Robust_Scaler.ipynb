# **RobustScaler**

### **Útil quando as variáveis possuem Outliers marginais**
* Ao contrário de **StandarScaler** que utiliza a **MÉDIA** em sua fórmula:

    **X'** *= ( X - Xmédia ) / Desvio Padrão* 

* O **RobustScaler** utiliza a **MEDIANA** em sua fórmula:

    **X'** *= ( X - Xmediana ) / IQR*

    **IQR** = Diferença Interquartílica (3º Quartil - 1º Quartil)

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
# Carga dos Dados
dados = pd.read_csv( 'pima-data.csv')
dados.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
dados['diabetes'] = encoder.fit_transform(dados['diabetes'])
dados.head()
```

```python
dados.dtypes
```

```python
dados.corr()['diabetes'].sort_values(ascending=False)
```

```python
X = dados.iloc[:, 0:9].drop(columns=['diab_pred', 'insulin', 'thickness', 'skin', 'diastolic_bp'], axis=1)
y = dados.iloc[:, 9]
```

```python
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.30, random_state=50)
```

```python
escaler = RobustScaler()
X_train_scaler = escaler.fit_transform(Xtrain)
X_test_scaler  = escaler.fit_transform(Xtest)
```

```python
from sklearn.neighbors import KNeighborsClassifier

modelo = KNeighborsClassifier()
modelo.fit(X_train_scaler, ytrain)

ypred = modelo.predict(X_test_scaler)
```

```python
print(classification_report(ytest, ypred))
```
